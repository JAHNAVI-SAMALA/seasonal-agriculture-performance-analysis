Missing values were investigated based on their distribution and seasonal characteristics. Since the project focuses on seasonal agricultural performance, appropriate seasonal-level imputation was considered to preserve differences between Kharif, Rabi and Zaid.

```python
# Fill Missing Yield
# Calculate yield from production and farm area
calculated_yield = (
    df["Production_Tonnes"] / df["Farm_Area_Hectares"]
)

# Fill only the missing yield values
df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(calculated_yield)

# Check remaining missing values
print("Missing Yield values:", df["Yield_Tonnes_Ha"].isnull().sum())
```

Missing yield values were derived using the relationship between production and cultivated area (Yield = Production / Farm Area), rather than using statistical imputation.

```python
# checking seasonal medians
print("Season-wise median rainfall:")
print(df.groupby("Season")["Rainfall_mm"].median())

print("\nSeason-wise median soil moisture:")
print(df.groupby("Season")["Soil_Moisture_pct"].median())
```

```python
# seasonal median imputation
# Fill missing rainfall using season-wise median
df["Rainfall_mm"] = df.groupby("Season")["Rainfall_mm"].transform(
    lambda x: x.fillna(x.median())
)

# Fill missing soil moisture using season-wise median
df["Soil_Moisture_pct"] = df.groupby("Season")["Soil_Moisture_pct"].transform(
    lambda x: x.fillna(x.median())
)
```

```python
# Outlier detection
Q1 = df["Yield_Tonnes_Ha"].quantile(0.25)
Q3 = df["Yield_Tonnes_Ha"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers = df[
    (df["Yield_Tonnes_Ha"] < lower_bound) |
    (df["Yield_Tonnes_Ha"] > upper_bound)
]

print("\nNumber of yield outliers:", len(outliers))
```

```python
outliers[[
    "Season",
    "Crop",
    "Farm_Area_Hectares",
    "Production_Tonnes",
    "Yield_Tonnes_Ha",
    "Profit_INR"
]].sort_values("Yield_Tonnes_Ha", ascending=False).head(20)
```

The initial IQR analysis identified 304 high-yield observations. Further investigation showed that 295 of these belonged to Sugarcane, indicating that the apparent outliers were primarily associated with crop-specific yield differences rather than random anomalous observations. Therefore, these observations were retained for analysis.

Drip irrigation recorded the highest average yield in Kharif and Rabi, while Sprinkler irrigation performed best in Zaid. This indicates that irrigation effectiveness varies across seasons, and no single irrigation method consistently produced the highest yield in every season.

Key Findings
Kharif recorded the highest average yield (5.63 tonnes/ha) and average profit (₹178,914.65).
Zaid recorded the lowest average yield (4.63 tonnes/ha) and a negative average profit (−₹24,804.82).
Sugarcane and Chilli were the most profitable crops, with average profits of approximately ₹8.17 lakh and ₹7.51 lakh, respectively.
Zaid recorded the highest average water usage (6,419.89 m³) and the lowest water-use efficiency (4.41).
Water-use efficiency showed a very strong positive correlation with yield (r = 0.92).
Revenue and profit also showed a strong positive correlation (r = 0.89).
Drip irrigation achieved the highest average yield in Kharif and Rabi, while Sprinkler performed best in Zaid.
Fertilizer, nitrogen, phosphorus, potassium and pesticide usage showed weak individual linear relationships with yield.
Crop-specific analysis showed that the apparent high-yield outliers were primarily associated with Sugarcane and were therefore retained.

### Recommendations

1. Promote efficient water management practices, particularly in seasons with high water demand.

2. Consider drip irrigation for Kharif and Rabi where appropriate, based on crop and farm conditions.

3. Evaluate sprinkler irrigation as an option for Zaid cultivation.

4. Give greater attention to crop selection because profitability varies considerably across crops and seasons.

5. Consider profitable crop-season combinations such as Chilli and Sugarcane while also accounting for their resource requirements.

6. Improve water-use efficiency rather than simply increasing water consumption.

7. Use data-driven seasonal planning to identify potentially low-profit or high-resource-use combinations before cultivation.

### Conclusion

The Seasonal Agriculture Performance Analysis provides insights into how agricultural productivity, profitability, environmental conditions, and resource utilization vary across seasons and crops. Kharif demonstrated the strongest overall performance, while Zaid showed lower productivity, negative average profitability, and lower water-use efficiency. Crop-level analysis revealed substantial differences in profitability, with Sugarcane and Chilli performing particularly well. The analysis also identified a very strong positive relationship between water-use efficiency and yield. Overall, the findings demonstrate the value of data-driven agricultural planning for selecting suitable crops, improving resource efficiency, and supporting more informed seasonal farming decisions.
